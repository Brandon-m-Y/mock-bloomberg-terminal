```python
import pandas as pd
import numpy as np
import requests
import os
from dotenv import load_dotenv, find_dotenv
from datetime import date, timedelta
```

```python
url = os.getenv("STOCK_ENDPOINT")
api_key = os.getenv("ALPHAVANTAGE_API_KEY")
stock = os.getenv("STOCK_NAME")
```

```python


params = {
    'function': 'TIME_SERIES_DAILY',
    'symbol': stock,
    'apikey': api_key
}
r = requests.get(url, params=params)
data = r.json()

print(data)
```

```python
daily_data = data['Time Series (Daily)']
```

```python

yesterday = (date.today() - timedelta(days=1)).isoformat()
day_before = (date.today() - timedelta(days=2)).isoformat()

yesterday_close = float(daily_data[yesterday]["4. close"])
day_before_close = float(daily_data[day_before]["4. close"])
```

```python
abs_val_diff = np.abs(yesterday_close - day_before_close)
abs_val_diff
```

```python
pct_diff = abs_val_diff / (yesterday_close + day_before_close/2) * 100
```

```python
news_api_key = os.getenv("NEWS_API_KEY")
```

```python
news_url = 'https://newsapi.org/v2/everything?'
 
params2 = {'q': 'tesla', 
       'from': '2026-10-01',
       'sortBy': 'popularity',
       'apiKey': news_api_key,
}
```

```python
r_news = requests.get(news_url, params=params2)
data = r_news.json()
```

```python
[a['content'] for a in data['articles'][:3]]
```

```python
import pandas as pd

df = pd.json_normalize(data['articles'])   # flattens source.id / source.name into columns
df[['source.name', 'title', 'publishedAt', 'content']].head(3)
```

```python
import re

def clean(text):
    if not text:
        return text
    text = re.sub(r'<[^>]+>', '', text)        # remove any HTML tags
    return re.sub(r'\s+', ' ', text).strip()    # collapse \r\n and extra spaces

df['content'] = df['content'].apply(clean)
```
